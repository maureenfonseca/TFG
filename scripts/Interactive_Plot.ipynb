# Short Term Analysis

This section contains the functions and process to plot time series for short term events (duration of few hours).

```python
# Paths for your data files
path1 = './Data/TURR_IMN.csv'
path2 = './Data/TURR_GPM_ST.csv'
```

```python
df = pd.read_csv(path1, sep=';')
```

```python
df
```

```python
# Create a dropdown widget for selecting resolution
resolution_options = ['1H', '3H', '6H', '12H'] 
resolution_dropdown = widgets.Dropdown(
    options=resolution_options,
    value='1H',
    description='Resolution:',
    disabled=False
)
```

```python
# Create an interactive plot
interactive_plot = widgets.interactive(
    plot_resol,    # plot_resol is a function imported from utils
    path1=widgets.fixed(path1),
    path2=widgets.fixed(path2),
    resolution=resolution_dropdown
)

# Display the interactive plot
display(interactive_plot)
```
